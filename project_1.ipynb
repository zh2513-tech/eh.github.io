# How Many People Are Injured in NYC Traffic Crashes?

## Introduction

Traffic accidents happen every day in New York City, but I was curious about how often they actually result in injuries. Are most crashes relatively minor, or do they tend to involve people getting hurt?

To explore this, I used NYC traffic collision data to look at the number of people injured per crash. I calculated the mean, median, and mode using pandas, then repeated the calculations using basic Python to see how the two approaches compare.

## Data Source

I used the [Motor Vehicle Collisions – Crashes dataset](https://data.cityofnewyork.us/d/h9gi-nx95) from NYC Open Data, which is maintained by the NYPD.

I filtered the original dataset to a selected period from November 1 through November 30, 2025, leaving 6,912 reported crashes. The filter does not include the entire calendar month. Each row represents one crash, and I focused on the `NUMBER OF PERSONS INJURED` column.

```python
import pandas as pd
```

```python
df = pd.read_csv("nyc_collisions.csv")

print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))

df.head()
```

## Descriptive Statistics Using Pandas

I started with pandas to calculate the mean, median, and mode of the number of people injured per crash.

The mean tells us the average, the median shows the middle value when the data is sorted, and the mode tells us which number appears most often. Looking at all three gives a better sense of what a typical crash looks like.

```python
injuries = df["NUMBER OF PERSONS INJURED"].dropna()

mean_pd = injuries.mean()
median_pd = injuries.median()
mode_pd = injuries.mode().tolist()

print("Mean:", round(mean_pd, 3))
print("Median:", median_pd)
print("Mode:", mode_pd)
```

### What Do These Results Mean?

The average number of people injured per crash was about 0.57, while both the median and mode were zero.

This suggests that crashes with no reported injuries were the most common. However, the average is higher than zero because some crashes involved one or more injured people.

## Calculating Statistics Without Pandas

Next, I calculated the same statistics without pandas or any built-in statistical functions.

I read the CSV file again using basic Python, then used loops, sorting, and a dictionary to calculate the mean, median, and mode manually. This took more code, but it helped me understand what pandas was doing behind the scenes.

```python
# Read the CSV using basic Python

import csv

values = []

with open("nyc_collisions.csv") as file:
    reader = csv.DictReader(file)

    for row in reader:
        if row["NUMBER OF PERSONS INJURED"] != "":
            values.append(int(row["NUMBER OF PERSONS INJURED"]))

# Calculate the mean

total = 0

for value in values:
    total += value

mean_basic = total / len(values)

print("Mean:", round(mean_basic, 3))

# Calculate the median

values.sort()

middle = len(values) // 2

if len(values) % 2 == 0:
    median_basic = (values[middle - 1] + values[middle]) / 2
else:
    median_basic = values[middle]

print("Median:", median_basic)

# Calculate the mode

counts = {}

for value in values:
    if value not in counts:
        counts[value] = 0

    counts[value] += 1

highest_count = max(counts.values())

modes_basic = []

for value in counts:
    if counts[value] == highest_count:
        modes_basic.append(value)

print("Mode:", modes_basic)
```

### Comparing the Two Methods

After running both sets of calculations, I compared the results to make sure they matched. All three statistics came out the same, which was a good way to check that my manual calculations worked correctly.

```python
print("Mean matches:", round(mean_pd, 3) == round(mean_basic, 3))
print("Median matches:", median_pd == median_basic)
print("Mode matches:", sorted(mode_pd) == sorted(modes_basic))
```

## Visualizing Injuries per Traffic Crash

I made a simple text-based bar chart to see how often crashes involved different numbers of injuries.

Each block represents approximately 100 crashes. Instead of using a plotting library, I used basic Python to create the bars with repeated symbols.

```python
print("Number of People Injured per NYC Traffic Crash")
print("November 2025")
print("Categories: Number of people injured per crash")
print("Bar length: Number of crashes")
print("Each █ represents about 100 crashes (minimum 1 block).\n")

for injury_count in sorted(counts):
    frequency = counts[injury_count]
    bar_length = round(frequency / 100)

    if frequency > 0 and bar_length == 0:
        bar_length = 1

    bar = "█" * bar_length

    print(f"{injury_count} injured | {bar} ({frequency:,} crashes)")
```

### What Does the Visualization Show?

The chart shows that crashes with no reported injuries were the most common, followed by crashes with one injured person. Crashes involving two or more injuries were much less frequent.

This also helps explain why the median and mode were both zero, even though the mean was about 0.57.

## Conclusion

Looking at the 6,912 traffic crashes in my dataset, I found that the average number of people injured per crash was about 0.57, while the median and mode were both zero. The bar chart made the pattern easier to see: most crashes had no reported injuries, but a considerable number still involved at least one injured person.

There are a few limitations to keep in mind. I only looked at a selected period in November 2025, so the findings might be different for other months. The dataset also only includes police-reported crashes.

This project was a useful way to practice working with real data. It also helped me understand how the mean, median, and mode can tell slightly different stories about the same dataset.
